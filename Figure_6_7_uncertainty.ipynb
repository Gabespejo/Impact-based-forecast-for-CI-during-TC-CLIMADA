```python
import numpy as np
import pandas as pd

hazard_stats = {
    'lead_time': [],
    'mean': [],
    'std': [],
    'min': [],
    'max': [],
    'cv_hazard': [],   # Coefficient of Variation
    'iqr': [],         # Interquartile Range
    'range': []        # Max - Min
}

# STEP 1: Collect all values across all lead times first for global normalization
global_all_data = []

for key, tc_obj in new_tc_names.items():
    for i in range(51):
        try:
            data = tc_obj.intensity[i].data
            if isinstance(data, np.ndarray):
                flat_data = data.flatten()
                flat_data = flat_data[~np.isnan(flat_data)]
                if flat_data.size > 0:
                    global_all_data.append(flat_data)
        except (IndexError, AttributeError, TypeError):
            continue

global_all_data = np.concatenate(global_all_data)
global_min = np.min(global_all_data)
global_max = np.max(global_all_data)

# STEP 2: Loop again and compute normalized stats per lead time using global min/max
for key, tc_obj in new_tc_names.items():
    ensemble_values = []

    for i in range(51):
        try:
            data = tc_obj.intensity[i].data
            if isinstance(data, np.ndarray):
                flat_data = data.flatten()
                flat_data = flat_data[~np.isnan(flat_data)]
                if flat_data.size > 0:
                    ensemble_values.append(flat_data)
        except (IndexError, AttributeError, TypeError) as e:
            print(f"Skipping member {i} for {key}: {e}")
            continue

    if ensemble_values:
        all_data = np.concatenate(ensemble_values)

        # Apply global normalization
        if global_max > global_min:
            all_data = (all_data - global_min) / (global_max - global_min)
        else:
            all_data = np.zeros_like(all_data)

        # Compute stats
        mean_val = np.mean(all_data)
        std_val = np.std(all_data)
        min_val = np.min(all_data)
        max_val = np.max(all_data)
        cv_val = std_val / mean_val if mean_val > 0 else np.nan
        q75, q25 = np.percentile(all_data, [75, 25])
        iqr_val = q75 - q25
        range_val = max_val - min_val

        # Store
        hazard_stats['lead_time'].append(key)
        hazard_stats['mean'].append(mean_val)
        hazard_stats['std'].append(std_val)
        hazard_stats['min'].append(min_val)
        hazard_stats['max'].append(max_val)
        hazard_stats['cv_hazard'].append(cv_val)
        hazard_stats['iqr'].append(iqr_val)
        hazard_stats['range'].append(range_val)
    else:
        print(f"No valid data for {key}, skipping.")

# Convert to DataFrame
df_hazard_stats = pd.DataFrame(hazard_stats)

# Optional: convert and sort by datetime if the key format includes a timestamp
try:
    df_hazard_stats['lead_time_dt'] = pd.to_datetime(
        df_hazard_stats['lead_time'].str.extract(r'(\d{8,})')[0], format='%Y%m%d%H%M%S')
    df_hazard_stats = df_hazard_stats.sort_values('lead_time_dt')
except Exception as e:
    print(f"Date parsing failed: {e}")

# Display result
print(df_hazard_stats)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
from matplotlib.lines import Line2D
from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.colorbar import ColorbarBase

# ====== CONFIG you can tweak ======
TRACK_ALPHA      = 0.5          # transparency of track lines
TRACK_COLOR_MAX  = 'red'
TRACK_COLOR_MIN  = '0.2'
TRACK_LW_MAX     = 1.5
TRACK_LW_MIN     = 1.5
TRACK_DASH_SEQ   = (6, 6)       # on-off dash pattern in points: 6 on, 6 off

TRACK_MAX_NO     = 11           # ensemble_number for max track (use 12 if attrs are 1-based)
TRACK_MIN_NO     = 3            # ensemble_number for min track (use 4  if attrs are 1-based)

XLIM = (32, 42)                 # crop longitude
YLIM = (-22, -10.5)             # crop latitude
TITLE = ""

# ---------- helpers ----------
def get_track_ds_by_ensemble_number(ens_container, ens_no):
    tracks = ens_container.get_track()
    for ds in tracks:
        try:
            if int(ds.attrs.get('ensemble_number', -999)) == int(ens_no):
                return ds
        except Exception:
            pass
    raise KeyError(f"Track with ensemble_number={ens_no} not found.")

def plot_track(ax, lon, lat, color, lw, alpha=TRACK_ALPHA, z=3):
    """
    Draw a *dashed* track using Line2D so the dashes definitely appear.
    NaNs in lon/lat will create breaks automatically.
    """
    lon = np.asarray(lon); lat = np.asarray(lat)
    ax.plot(
        lon, lat,
        color=color,
        linewidth=lw,
        alpha=alpha,
        zorder=z,
        linestyle=(0, TRACK_DASH_SEQ),  # true dashed pattern
        solid_capstyle='butt',
        dash_capstyle='butt',
    )

# ---------- data ----------
lons = imph2['1 day'].coord_exp[:, 1]
lats = imph2['1 day'].coord_exp[:, 0]
M = imph2['1 day'].imp_mat.toarray()

totals_per_member = M.sum(axis=1)
min_idx = int(np.argmin(totals_per_member))
max_idx = int(np.argmax(totals_per_member))
delta = M[max_idx] - M[min_idx]        # difference you want to show

moz = gadm_MOZ_gdf.to_crs("EPSG:4326")
gdf = gpd.GeoDataFrame({"delta": delta},
                       geometry=gpd.points_from_xy(lons, lats),
                       crs="EPSG:4326")

# ---------- discrete 8-color scheme with 0.05 steps (0–0.35, plus ≥0.35) ----------
# 8 bins => 9 boundaries (last open-ended)
bounds = [0.00, 0.05, 0.10, 0.15, 0.20, 0.25, 0.30, 0.35, np.inf]
colors = [
    "#fff7bc",  # 0.00–0.05 (very light yellow)
    "#fee391",  # 0.05–0.10
    "#fec44f",  # 0.10–0.15
    "#fdb863",  # 0.15–0.20
    "#fb6a4a",  # 0.20–0.25
    "#fc4e2a",  # 0.25–0.30
    "#ef3b2c",  # 0.30–0.35
    "#99000d",  # ≥0.35
]
cmap = ListedColormap(colors, name="delta_8step")
norm = BoundaryNorm(bounds, ncolors=len(colors), clip=False)

# ---------- plot base map + delta points ----------
fig, ax = plt.subplots(figsize=(10, 10))
moz.plot(ax=ax, color='white', edgecolor='black', zorder=0)

gdf.plot(ax=ax, column='delta', cmap=cmap, norm=norm,
         markersize=20, alpha=0.85, legend=False, zorder=1)

# crop + cosmetic axes + optional title
ax.set_xlim(XLIM)
ax.set_ylim(YLIM)
ax.set_xlabel(""); ax.set_ylabel("")
ax.set_xticks([]); ax.set_yticks([])
ax.set_title(TITLE, fontsize=14, pad=10)

# ---------- overlay the two ensemble tracks (truly dashed) ----------
ens_key = 'ens_FREDDY_20230310120000'
ens = ens_variables[ens_key]

ds_max = get_track_ds_by_ensemble_number(ens, TRACK_MAX_NO)
ds_min = get_track_ds_by_ensemble_number(ens, TRACK_MIN_NO)

lon_max, lat_max = ds_max['lon'].values, ds_max['lat'].values
lon_min, lat_min = ds_min['lon'].values, ds_min['lat'].values

plot_track(ax, lon_max, lat_max, color=TRACK_COLOR_MAX, lw=TRACK_LW_MAX, alpha=TRACK_ALPHA, z=3)
plot_track(ax, lon_min, lat_min, color=TRACK_COLOR_MIN, lw=TRACK_LW_MIN, alpha=TRACK_ALPHA, z=3)

# ---------- save map ----------
plt.tight_layout()
out_map = "/cluster/home/gespejo/MOZ_struc_plots_Freddy/Delta_impact_max_minus_min_scenario.jpg"
plt.savefig(out_map, dpi=1500, bbox_inches='tight', format='jpg')
plt.show()
```
