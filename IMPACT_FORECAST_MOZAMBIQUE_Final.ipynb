```python
from datetime import datetime, timedelta
import logging
import os

logging.getLogger('climada.hazard.tc_tracks_forecast').setLevel(logging.WARNING)

def fetch_multiple_ecmwf_data(base_folder, start_date, end_date, event_name):
    current_date = start_date
    forecast_variables = {}  # Dictionary to store forecast variables

    while current_date <= end_date:
        for utc_hour in [0,12]:  # Only check 0 UTC and 12 UTC
            date_folder = current_date.replace(hour=utc_hour, minute=0, second=0).strftime("%Y%m%d%H%M%S")
            folder_path = os.path.join(base_folder, date_folder)

            if os.path.exists(folder_path):
                forecast = TCForecast()
                files_to_fetch = []
                for root, dirs, files in os.walk(folder_path):
                    for file in files:
                        if event_name in file:
                            files_to_fetch.append(os.path.join(root, file))

                if files_to_fetch:
                    print(f"Fetching data for {date_folder}")
                    forecast.fetch_ecmwf(files=files_to_fetch)
                    variable_name = f"{event_name}_{date_folder}"
                    forecast_variables[variable_name] = forecast
                else:
                    print(f"No data found for {date_folder}")
            else:
                print(f"Folder not found: {date_folder}")

        current_date += timedelta(days=1)

    return forecast_variables
```

```python
def _find_duplicates(overlap_gdf):
    indices_dupl = np.unique([index_right for index_right, index_left in zip(overlap_gdf.index_right, overlap_gdf.index_left)
                    if index_right != index_left])
    indices_dupl = indices_dupl[:int(len(indices_dupl/2))] # only first half of duplicates should be removed!
    return indices_dupl

def _remove_duplicates(orig_gdf, indices_dupl):
    indices_keep = list(set(orig_gdf.index).difference(set(indices_dupl)))
    return orig_gdf.loc[indices_keep]

def find_remove_duplicates(orig_gdf):
    if 'index' in orig_gdf.columns:
        orig_gdf = orig_gdf.drop('index', axis=1)
    overlap_gdf = orig_gdf.sjoin(orig_gdf)
    if 'index_left' not in overlap_gdf.columns:
        overlap_gdf = overlap_gdf.reset_index()
        overlap_gdf = overlap_gdf.rename({'index':'index_left'}, axis=1)  
    indices_dupl = _find_duplicates(overlap_gdf)
    return _remove_duplicates(orig_gdf, indices_dupl) 
```

## CLASSIFICATION OF THE HOSPITAL BY LEVELS 

```python
# Define a function to determine the level based on the criteria
# Suppose you have a GeoDataFrame named 'gdf' with columns 'amenity', 'building', 'healthcare', 'healthcare__hospital', 'healthcare__clinic', 'healthcare__doctor'

# Define a function to determine the level based on the criteria
def determine_level(row):
    if row['amenity'] == 'hospital':
        return 4
    elif row['amenity'] == 'clinic':
        return 2
    elif row['amenity'] == 'doctors':
        return 1
    elif row['building'] == 'hospital':
        return 4
    elif row['building'] == 'clinic':
        return 2
    elif row['building'] == 'doctor':
        return 1
    elif row['healthcare'] == 'hospital':
        return 4
    elif row['healthcare'] == 'clinic':
        return 2
    elif row['healthcare'] == 'doctor':
        return 1
    elif row['healthcare__hospital']:
        return 4
    elif row['healthcare__clinic']:
        return 2
    elif row['healthcare__doctor']:
        return 1
    else:
        return None
```

```python
import geopandas as gpd
from shapely.geometry import Point

def create_points_geodfs_from_dict(ed_dict, levels_gdf, crs):
    """
    Create a dictionary of GeoDataFrames from a dictionary of coordinates, adding 'levels' from a GeoDataFrame.

    Parameters:
    ed_dict (dict): Dictionary where keys are identifiers and values contain coord_exp attribute.
    levels_gdf (GeoDataFrame): GeoDataFrame containing the 'levels' column to be added.
    crs: Coordinate reference system to use for the GeoDataFrames.

    Returns:
    dict: A dictionary of GeoDataFrames with 'levels' column added.
    """
    geodfs = {}
    for key, impact in ed_dict.items():
        coords = impact.coord_exp
        levels = levels_gdf['levels']

        # Creating GeoDataFrame with geometry and 'levels'
        points = [Point(lon, lat) for lat, lon in coords]
        points_gdf = gpd.GeoDataFrame({'geometry': points, 'levels': levels}, crs=crs)

        geodfs[key] = points_gdf

    return geodfs
```

```python
import geopandas as gpd

def convert_col_to_gdfs(df, substring):
    """
    Convert columns from a DataFrame to separate GeoDataFrames based on a substring.

    Parameters:
    - df: DataFrame containing the columns to convert
    - substring: string to match in column names to select them for conversion

    Returns:
    - gdf_imp_pp: dictionary of GeoDataFrames, each derived from a column in df
    """
    # List of names that you want to convert to GeoDataFrames
    column_names = [col_name for col_name in df.columns if substring in col_name]

    # Dictionary to hold the created GeoDataFrames
    gdf_imp_pp = {}

    for col_name in column_names:
        # Create a new DataFrame containing only the relevant columns
        sub_df = df[['name','latitude', 'longitude', 'value', 'geometry','levels', col_name]]
        
        # Convert the DataFrame to a GeoDataFrame using the existing 'geometry' column
        gdf = gpd.GeoDataFrame(sub_df, geometry='geometry', crs='EPSG:4326')
        
        # Store the GeoDataFrame in the dictionary
        gdf_imp_pp[col_name] = gdf
    
    return gdf_imp_pp
```

```python
import geopandas as gpd

def sjoin_gdfs(gdf_imp_pp_dict, other_gdf, op='within'):
    """
    Perform a spatial join for each GeoDataFrame in a dictionary with another GeoDataFrame.
    Join the Geodataframes of the impact of the population with the geodataframe with the 
    administrative divisions of a country, in this case Mozambique,it is needed a shapefile. 
    
    Parameters:
    - gdfs_dict: dictionary of GeoDataFrames to be joined
    - other_gdf: the other GeoDataFrame to join with
    - op: binary predicate passed to sjoin, default 'within'
    
    Returns:
    - joined_gdfs: dictionary of GeoDataFrames resulting from the spatial joins
    """
    joined_gdfs_imp_pp = {}
    for key, gdf in gdf_imp_pp_dict.items():
        joined_gdfs_imp_pp[key] = gpd.sjoin(gdf, other_gdf, op=op,how='left')
    
    return joined_gdfs_imp_pp
```

```python
def hfacilities_impsum_level(gdfs_dict):
    """
    Processes a dictionary of GeoDataFrames to sum the values by province and level and keeps only sums above 0.50.
    
    Args:
    gdfs_dict (dict): A dictionary where each key is a string and each value is a GeoDataFrame.
    
    Returns:
    dict: A dictionary where each key is the original key from gdfs_dict and each value is a DataFrame with summed values above 0.50.
    """
    filtered_sums_dict = {}

    for key, gdf in gdfs_dict.items():
        # Identify the column that starts with 'imph1'
        imp_col = [col for col in gdf.columns if col.startswith('imph')][0]

        # Group by 'NAME_1' and 'levels', then sum the imp_col values
        grouped = gdf.groupby(['NAME_1', 'levels'])
        sums = grouped[imp_col].sum().reset_index(name='Total_Sum')

        # Keep only the rows where 'Total_Sum' is greater than 0.50
        filtered_sums = sums[sums['Total_Sum'] > 0.5]

        # Store the result in the dictionary
        filtered_sums_dict[key] = filtered_sums

    return filtered_sums_dict

# Example usage:
nh2 = hfacilities_impsum_level(jd_gdfs_imph2)
```

```python
import pandas as pd

def hfacilities_level_nfailure(gdfs_dict):
    """
    Processes a dictionary of GeoDataFrames to sum the values and count the occurrences of facilities by province and level.
    
    Args:
    gdfs_dict (dict): A dictionary where each key is a string and each value is a GeoDataFrame.
    
    Returns:
    dict: A dictionary where each key is the original key from gdfs_dict, and each value is a DataFrame with summed values and counts.
    """
    counts_dict = {}

    for key, gdf in gdfs_dict.items():
        # Identify the column that starts with 'imph1'
        imp_col = [col for col in gdf.columns if col.startswith('imph')][0]

        # Filter rows where 'imp_col' is greater than 0
        filtered_data = gdf[gdf[imp_col] > 0]

        # Group by 'NAME_1' and 'levels', then sum the imp_col values and count occurrences
        grouped = filtered_data.groupby(['NAME_1', 'levels'])
        sums = grouped[imp_col].sum().reset_index(name='Total_Sum')
        counts = grouped.size().reset_index(name='Count')

        # Merging the sums and counts
        merged_df = pd.merge(sums, counts, on=['NAME_1', 'levels'])

        # Store the result in the dictionary
        counts_dict[key] = merged_df

    return counts_dict
```

```python
import geopandas as gpd

def aggregate_impact_by_regions_for_all(points_geodfs, admin_regions_gdf, ed1, admin_col_name='NAME_1'):
    """
    Aggregate impact by administrative regions for all keys in points_geodfs.

    Parameters:
    points_geodfs (dict): Dictionary of GeoDataFrames containing point geometries.
    admin_regions_gdf (geopandas.GeoDataFrame): GeoDataFrame containing administrative region geometries.
    ed1 (dict): Dictionary containing impact objects with impact_at_reg method.
    admin_col_name (str): Column name in admin_regions_gdf that contains region names. Default is 'NAME_1'.

    Returns:
    dict: A dictionary of DataFrames with aggregated impact data for each key.
    """
    aggregated_impacts = {}
    for key, points_gdf in points_geodfs.items():
        # Spatial join to find which admin region each point falls within
        joined_gdf = gpd.sjoin(points_gdf, admin_regions_gdf[[admin_col_name, 'geometry']], how='left', op='within')

        # Use the result of the spatial join as aggregation regions in impact_at_reg
        if key in ed1:
            aggregated_df = ed1[key].impact_at_reg(agg_regions=joined_gdf[admin_col_name].tolist())

            # Remove 'nan' column if it exists
            if 'nan' in aggregated_df.columns:
                aggregated_df.drop(columns=['nan'], inplace=True)

            aggregated_impacts[key] = aggregated_df
        else:
            print(f"Key '{key}' not found in ed1.")

    return aggregated_impacts
```

```python
def seaborn_whiskers(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    upper_whisker = group[group <= q3 + 1.5 * iqr].max()
    lower_whisker = group[group >= q1 - 1.5 * iqr].min()
    return pd.Series({'lower_whisker': lower_whisker, 'upper_whisker': upper_whisker})
```

```python
def extract_columns_from_exposures(exp_instance, date_forecast):
    try:
        # Find a column that ends with the provided date_forecast.
        matching_columns = [col for col in exp_instance.gdf.columns if col.endswith(date_forecast)]
        
        if not matching_columns:
            print(f"No column ending with {date_forecast} found.")
            return None, None, None, None
        
        # Taking the first matching column, but you may adjust as needed.
        impact_column = matching_columns[0]

        # Extract columns - adjust if your data structure requires.
        longitude = exp_instance.gdf['longitude']
        latitude = exp_instance.gdf['latitude']
        #levels = exp_instance.gdf['levels']
        levels = exp_instance.gdf[['levels','latitude', 'longitude']]
        impact = exp_instance.gdf[impact_column]  # Adjusted this line to use variable impact_column
    except KeyError as ke:
        print(f"Error: {ke}. Could not extract data. Check column names and data structure.")
        return None, None, None, None
    
    return longitude, latitude, levels, impact
```

```python
from datetime import datetime
import pandas as pd

def extract_points_to_plot(exposures_dict, exp_key, date_forecast, event_date):
    try:
        dt_forecast = datetime.strptime(date_forecast, "%Y%m%d%H%M%S")
        dt_event = datetime.strptime(event_date, "%Y%m%d%H%M%S")
    except ValueError as ve:
        print(f"Error: {ve}. Ensure date_forecast and event_date are correctly formatted.")
        return None
    
    if exp_key not in exposures_dict:
        raise KeyError(f"{exp_key} not found in exposures_dict.")
    
    lead_time = (dt_event - dt_forecast).days
    
    if lead_time < 0:
        print(f"Warning: The event date {event_date} is earlier than the forecast date {date_forecast}. Lead time: {lead_time} days.")
    elif lead_time == 0:
        print(f"Note: The event date {event_date} and forecast date {date_forecast} are the same. Lead time: {lead_time} days.")
    else:
        print(f"Note: Calculating with lead time of {lead_time} days.")
        
    exp_instance = exposures_dict[exp_key]
    
    # No need to build the impact column based on exp_key, we’ll directly use date_forecast.
    
    exp_lon, exp_lat, exp_levels, exp_impact = extract_columns_from_exposures(exp_instance, date_forecast)
    
    return {
        'longitude': exp_lon,
        'latitude': exp_lat,
        'levels': exp_levels,
        'impact': exp_impact,
        'lead_time': lead_time
    }
```

```python
import geopandas as gpd

def convert_col_to_gdfs(df, substring):
    """
    Convert columns from a DataFrame to separate GeoDataFrames based on a substring.

    Parameters:
    - df: DataFrame containing the columns to convert
    - substring: string to match in column names to select them for conversion

    Returns:
    - gdf_imp_pp: dictionary of GeoDataFrames, each derived from a column in df
    """
    # List of names that you want to convert to GeoDataFrames
    column_names = [col_name for col_name in df.columns if substring in col_name]

    # Dictionary to hold the created GeoDataFrames
    gdf_imp_pp = {}

    for col_name in column_names:
        # Create a new DataFrame containing only the relevant columns
        sub_df = df[['latitude', 'longitude', 'value', 'geometry', col_name]]
        
        # Convert the DataFrame to a GeoDataFrame using the existing 'geometry' column
        gdf = gpd.GeoDataFrame(sub_df, geometry='geometry', crs='EPSG:4326')
        
        # Store the GeoDataFrame in the dictionary
        gdf_imp_pp[col_name] = gdf
    
    return gdf_imp_pp
```

```python
import geopandas as gpd

def sjoin_gdfs(gdf_imp_pp_dict, other_gdf, op='within'):
    """
    Perform a spatial join for each GeoDataFrame in a dictionary with another GeoDataFrame.
    Join the Geodataframes of the impact of the population with the geodataframe with the 
    administrative divisions of a country, in this case Mozambique,it is needed a shapefile. 
    
    Parameters:
    - gdfs_dict: dictionary of GeoDataFrames to be joined
    - other_gdf: the other GeoDataFrame to join with
    - op: binary predicate passed to sjoin, default 'within'
    
    Returns:
    - joined_gdfs: dictionary of GeoDataFrames resulting from the spatial joins
    """
    joined_gdfs_imp_pp = {}
    for key, gdf in gdf_imp_pp_dict.items():
        joined_gdfs_imp_pp[key] = gpd.sjoin(gdf, other_gdf, op=op)
    
    return joined_gdfs_imp_pp
```

```python
def find_ppaffected_adm1 (gdfs_dict, column_to_groupby='NAME_1'):
    """
    Group by a column and sum another column for all GeoDataFrames in a dictionary.

    Parameters:
    - gdfs_dict: dictionary of GeoDataFrames
    - column_to_groupby: string, name of the column to group by
    
    Returns:
    - pp_data: dictionary containing the grouped and summed DataFrames, with the same keys as the input
    """
    pp_admin1 = {}
    
    for track_time, gdf in gdfs_dict.items():
        # Extracting the relevant column name from the dictionary key
        column_to_sum = track_time
        
        # Grouping by 'NAME_2' and summing the relevant column, then resetting index
        grouped_df = gdf.groupby(column_to_groupby)[column_to_sum].sum().reset_index().round()
        
        # Storing the resulting DataFrame in the new dictionary with the same key
        pp_admin1[track_time] = grouped_df
    
    return pp_admin1
```

```python
def filter_pp_above_zero(data_dict):
    """
    Filter the rows where the key column is above 0 in all DataFrames in the input dictionary.

    Parameters:
    - data_dict: dictionary of DataFrames
    
    Returns:
    - filtered_data: dictionary containing the filtered DataFrames, with the same keys as the input
    """
    ppdata_abzero = {}
    
    for key, df in data_dict.items():
        # Check if the key exists as a column in the DataFrame
        if key not in df.columns:
            print(f"Column {key} not found in DataFrame associated with key {key}.")
            continue

        # Filter the DataFrame where values in the key column are > 0
        filtered_df = df[df[key] > 0].copy()
        ppdata_abzero[key] = filtered_df
    
    return ppdata_abzero
```

```python
def subset_failure_above20(dataframe_dict, key, start):
    # Extract dataframe based on key
    df = dataframe_dict[key]
    
    # Extracting the date columns automatically based on the naming pattern
    date_columns = [col for col in df.columns if col.startswith(f'{start}_MOZ_new_FREDDY_')]
    
    #print(f"Found {len(date_columns)} columns matching pattern '{imph2}_MOZ_new_FREDDY_'.")  # Diagnostic print

    # Columns to keep for each new dataframe
    columns_to_keep = ['name', 'geometry', 'levels', 'latitude', 'longitude']

    # Creating a dictionary to store new dataframes
    new_dfs = {}
    
    # Iterating over date columns and creating new dataframes
    for date_col in date_columns:
        # Extracting year, month, and day from the column name to create a key
        _, date_str = date_col.rsplit('_', 1)
        year, month, day = date_str[:4], date_str[4:6], date_str[6:8]
        
        # Creating a key for each dataframe in 'YYYY-MM-DD' format
        new_key = f'disrupt_{year}-{month}-{day}'
        
        # Filtering the dataframe where current column > 0.20
        filtered_df = df[df[date_col] > 0.20]
        
        print(f"Found {len(filtered_df)} rows in column '{date_col}' with values > 0.20.")  # Diagnostic print
        
        # Keeping only the necessary columns and adding it to the new_dfs dictionary
        new_dfs[new_key] = filtered_df[columns_to_keep + [date_col]]

    return new_dfs

# Usage
disrupt2_dfs = subset_failure_above20(joined_dfs, 'imph_2', 'imph2')
```

```python
import geopandas as gpd

def calculate_nearest_population(gdf, pop_gdf):
    """
    Calculate the nearest population point for each entry in a GeoDataFrame.

    Parameters:
    - gdf: GeoDataFrame where nearest population will be calculated for each entry.
    - pop_gdf: GeoDataFrame representing population data.

    Returns:
    - gdf: Updated GeoDataFrame with nearest population index and affected population.
    """
    # Calculate the index of the nearest population point for each entry in gdf
    gdf['nearest_pop_index'] = gdf.geometry.apply(lambda x: pop_gdf.distance(x).idxmin())
    
    # Create a new column to store the affected population for each entry in gdf
    gdf['affected_population'] = gdf['nearest_pop_index'].apply(lambda idx: pop_gdf.at[idx, 'value'])
    
    return gdf

# Apply the function to every GeoDataFrame in geodisrupt2_dfs
for key, gdf in geodisrupt2_dfs.items():
    geodisrupt2_dfs[key] = calculate_nearest_population(gdf, gdf_worldpop_MOZ.gdf)
```
