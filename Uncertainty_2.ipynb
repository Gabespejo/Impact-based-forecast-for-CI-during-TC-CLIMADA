```python
import numpy as np
import pandas as pd

hazard_stats = {
    'lead_time': [],
    'mean': [],
    'std': [],
    'min': [],
    'max': [],
    'cv_hazard': [],   # Coefficient of Variation
    'iqr': [],         # Interquartile Range
    'range': []        # Max - Min
}

# STEP 1: Collect all values across all lead times first for global normalization
global_all_data = []

for key, tc_obj in new_tc_names.items():
    for i in range(51):
        try:
            data = tc_obj.intensity[i].data
            if isinstance(data, np.ndarray):
                flat_data = data.flatten()
                flat_data = flat_data[~np.isnan(flat_data)]
                if flat_data.size > 0:
                    global_all_data.append(flat_data)
        except (IndexError, AttributeError, TypeError):
            continue

global_all_data = np.concatenate(global_all_data)
global_min = np.min(global_all_data)
global_max = np.max(global_all_data)

# STEP 2: Loop again and compute normalized stats per lead time using global min/max
for key, tc_obj in new_tc_names.items():
    ensemble_values = []

    for i in range(51):
        try:
            data = tc_obj.intensity[i].data
            if isinstance(data, np.ndarray):
                flat_data = data.flatten()
                flat_data = flat_data[~np.isnan(flat_data)]
                if flat_data.size > 0:
                    ensemble_values.append(flat_data)
        except (IndexError, AttributeError, TypeError) as e:
            print(f"Skipping member {i} for {key}: {e}")
            continue

    if ensemble_values:
        all_data = np.concatenate(ensemble_values)

        # Apply global normalization
        if global_max > global_min:
            all_data = (all_data - global_min) / (global_max - global_min)
        else:
            all_data = np.zeros_like(all_data)

        # Compute stats
        mean_val = np.mean(all_data)
        std_val = np.std(all_data)
        min_val = np.min(all_data)
        max_val = np.max(all_data)
        cv_val = std_val / mean_val if mean_val > 0 else np.nan
        q75, q25 = np.percentile(all_data, [75, 25])
        iqr_val = q75 - q25
        range_val = max_val - min_val

        # Store
        hazard_stats['lead_time'].append(key)
        hazard_stats['mean'].append(mean_val)
        hazard_stats['std'].append(std_val)
        hazard_stats['min'].append(min_val)
        hazard_stats['max'].append(max_val)
        hazard_stats['cv_hazard'].append(cv_val)
        hazard_stats['iqr'].append(iqr_val)
        hazard_stats['range'].append(range_val)
    else:
        print(f"No valid data for {key}, skipping.")

# Convert to DataFrame
df_hazard_stats = pd.DataFrame(hazard_stats)

# Optional: convert and sort by datetime if the key format includes a timestamp
try:
    df_hazard_stats['lead_time_dt'] = pd.to_datetime(
        df_hazard_stats['lead_time'].str.extract(r'(\d{8,})')[0], format='%Y%m%d%H%M%S')
    df_hazard_stats = df_hazard_stats.sort_values('lead_time_dt')
except Exception as e:
    print(f"Date parsing failed: {e}")

# Display result
print(df_hazard_stats)
```

```python
import matplotlib.pyplot as plt
import geopandas as gpd
import numpy as np
import matplotlib.colors as mcolors

# Extract coordinates (ensuring correct order)
longitudes = imph2['1 day'].coord_exp[:, 1]  # Longitude (X)
latitudes = imph2['1 day'].coord_exp[:, 0]   # Latitude (Y)

# Extract impact matrix and select specific scenarios
impact_matrix = imph2['1 day'].imp_mat.toarray()
selected_scenarios = [4, 12]  # Two scenarios

# Ensure Mozambique boundary is loaded
if 'gadm_MOZ_gdf' in globals():
    mozambique_gdf = gadm_MOZ_gdf.to_crs("EPSG:4326")  # Convert to WGS84
else:
    raise ValueError("gadm_MOZ_gdf is not available. Please load the Mozambique boundary dataset.")

# Define the colorbar levels and colors
color_ranges = [0, 0.02, 0.10, 0.20, 0.50, 1.0]
colors = [
    (204 / 255, 255 / 255, 102 / 255, 0.1),   # Light green (No Damage)
    (204 / 255, 255 / 255, 102 / 255, 0.9),   # Strong green (Minor Impact)
    (255 / 255, 255 / 255, 0 / 255, 0.9),     # Yellow (Major Impact)
    (255 / 255, 153 / 255, 0 / 255, 0.9),     # Orange (Severe Impact)
    (255 / 255, 0 / 255, 0 / 255, 0.9)        # Red (Destroyed)
]
my_cmap = mcolors.ListedColormap(colors)
norm = mcolors.BoundaryNorm(color_ranges, my_cmap.N)

# Create 1x2 subplot layout for 2 scenarios
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Plot each selected scenario
for i, scenario in enumerate(selected_scenarios):
    ax = axes[i]  # Adjusted for 1x2 grid

    # Plot Mozambique boundary
    mozambique_gdf.plot(ax=ax, color='white', edgecolor='black')

    # Create GeoDataFrame for the selected scenario's impact data
    scenario_gdf = gpd.GeoDataFrame(
        {'Impact': impact_matrix[scenario]},
        geometry=gpd.points_from_xy(longitudes, latitudes),
        crs="EPSG:4326"
    )

    # Plot impact points with the correct colormap
    scenario_gdf.plot(
        ax=ax, column='Impact', cmap=my_cmap, markersize=20, alpha=0.7,
        legend=False, norm=norm
    )

    # Set plot limits and titles
    ax.set_xlim([32, 42])
    ax.set_ylim([-22, -13.5]) 
    ax.set_title(f"Scenario {scenario} Impact", fontsize=16)
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")

# Add shared colorbar for both subplots
cbar_ax = fig.add_axes([0.92, 0.2, 0.02, 0.6])
sm = plt.cm.ScalarMappable(cmap=my_cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, cax=cbar_ax, boundaries=color_ranges, ticks=color_ranges)
cbar.set_label("Impact Value", fontsize=12)

# Adjust the layout and save the figure
plt.tight_layout(rect=[0, 0, 0.9, 1])
output_path = "/cluster/home/gespejo/MOZ_struc_plots_Freddy/low_high_impact_scenarios_Freddy_20231003120000.jpg"
plt.savefig(output_path, dpi=600, bbox_inches='tight', format='jpg')  # Saving the figure

print(f"Plot saved successfully at: {output_path}")
plt.show()
```
