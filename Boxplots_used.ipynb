```python
def seaborn_whiskers(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    upper_whisker = group[group <= q3 + 1.5 * iqr].max()
    lower_whisker = group[group >= q1 - 1.5 * iqr].min()
    return pd.Series({'lower_whisker': lower_whisker, 'upper_whisker': upper_whisker})
```

```python
import geopandas as gpd
from shapely.geometry import Point

def create_points_geodfs_hfacilities(ed_dict, levels_gdf, crs):
    """
    Create a dictionary of GeoDataFrames from a dictionary of coordinates, adding 'levels' from a GeoDataFrame.

    Parameters:
    ed_dict (dict): Dictionary where keys are identifiers and values contain coord_exp attribute.
    levels_gdf (GeoDataFrame): GeoDataFrame containing the 'levels' column to be added.
    crs: Coordinate reference system to use for the GeoDataFrames.

    Returns:
    dict: A dictionary of GeoDataFrames with 'levels' column added.
    """
    geodfs = {}
    for key, impact in ed_dict.items():
        coords = impact.coord_exp
        levels = levels_gdf[['name','levels','latitude','longitude']]

        # Creating GeoDataFrame with geometry and 'levels'
        points = [Point(lon, lat) for lat, lon in coords]
        points_gdf = gpd.GeoDataFrame({'name':name,'geometry': points, 'levels': levels,'latitude':latitude,
                                       'longitud':longitud}, crs=crs)

        geodfs[key] = points_gdf

    return geodfs
```

```python
def scale_imp_pp(gdf_dict, substring, divisor_column):
    """
    Add a new column to each GeoDataFrame in the dictionary by dividing a target column by the divisor column on a row-by-row basis.

    Parameters:
    - gdf_dict: Dictionary of GeoDataFrames.
    - substring: Substring to identify the target column whose name starts with this substring.
    - divisor_column: Name of the column to divide the target column by.

    Returns:
    - Modified dictionary of GeoDataFrames with the new column added.
    """
    for key, gdf in gdf_dict.items():
        # Find the column that starts with the specified substring
        target_col = [col for col in gdf.columns if col.startswith(substring)][0]

        # Perform the division for each row and create a new column
        gdf['imp_scale'] = gdf.apply(lambda row: row[target_col] / row[divisor_column] if row[divisor_column] != 0 else None, axis=1)

        # Update the dictionary with the modified GeoDataFrame
        gdf_dict[key] = gdf

    return gdf_dict

# Usage example
jd_gdfs_pp1 = scale_imp_pp(jd_gdfs_pp1, 'impp1_MOZ_FREDDY_', 'value')
```

```python
def calculate_nearest_population(geodisrupt_gdf, pop_gdf):
    """
    Calculate the nearest population point for each entry in GeoDataFrames within geodisrupt2_dfs.

    Parameters:
    - geodisrupt_gdf: Dictionary of GeoDataFrames where nearest population will be calculated for each entry.
    - pop_gdf: GeoDataFrame representing population data.

    Returns:
    - geodisrupt_gdf: Dictionary of updated GeoDataFrames with nearest population index and affected population.
    """
    for key, gdf in geodisrupt_gdf.items():
        # Calculate the index of the nearest population point for each entry in gdf
        gdf['nearest_pop_index'] = gdf.geometry.apply(lambda x: pop_gdf.distance(x).idxmin())
        
        # Create a new column to store the affected population for each entry in gdf
        gdf['affected_population'] = gdf['nearest_pop_index'].apply(lambda idx: pop_gdf.at[idx, 'people'])
        
        # Convert the 'affected_population' column to a numeric type, round to integers without decimals, and cast to int
        gdf['affected_population'] = pd.to_numeric(gdf['affected_population']).round(0).astype(int)
        
        # Update the dictionary with the calculated GeoDataFrame
        geodisrupt_gdf[key] = gdf
    
    return geodisrupt_gdf
```

```python
def cal_nearestpop_disrup(geodisrupt_gdf, disrupt_data_gdf):
    """
    Calculate the nearest population point for each entry in GeoDataFrames within geodisrupt_gdf
    using the corresponding GeoDataFrames in disrup_pp1.

    Parameters:
    - geodisrupt_gdf: Dictionary of GeoDataFrames where nearest population will be calculated for each entry.
    - disrup_pp1: Dictionary of GeoDataFrames representing population data for different days.

    Returns:
    - geodisrupt_gdf: Dictionary of updated GeoDataFrames with nearest population index and affected population.
    """
    for key in geodisrupt_gdf:
        if key in disrupt_data_gdf:
            # Access the corresponding population GeoDataFrame
            pop_gdf = disrupt_data_gdf[key]

            # Calculate the index of the nearest population point for each entry in geodisrupt_gdf
            geodisrupt_gdf[key]['nearest_pop_index'] = geodisrupt_gdf[key].geometry.apply(lambda x: pop_gdf.distance(x).idxmin())

            # Create a new column to store the affected population for each entry in geodisrupt_gdf
            geodisrupt_gdf[key]['pp_affected'] = geodisrupt_gdf[key]['nearest_pop_index'].apply(lambda idx: pop_gdf.at[idx, 'imp_people'])

            # Convert the 'affected_population' column to a numeric type, round to integers without decimals, and cast to int
            geodisrupt_gdf[key]['pp_affected'] = pd.to_numeric(geodisrupt_gdf[key]['pp_affected']).round(0).astype(int)

    return geodisrupt_gdf
```

```python
import geopandas as gpd
from geopandas.tools import sjoin

def process_time_cases(disrupt_rd1, disrup_pp1):
    # Buffer distance for 2 km in degrees
    buffer_distance_2km_deg = 2 / 111  # 2 km converted to degrees

    # Dictionary to store results
    results_dict = {}

    # Assuming your data is already loaded into disrupt_rd1 and disrup_pp1
    for time_case in disrupt_rd1.keys():
        # 1. Buffer the Linestrings
        disrupt_rd1[time_case]['buffered_geometry'] = disrupt_rd1[time_case].geometry.buffer(buffer_distance_2km_deg)

        # Set the buffered geometry as the active geometry
        disrupt_rd1[time_case].set_geometry('buffered_geometry', inplace=True)

        # Rename the existing 'index_right' column to avoid conflict
        disrup_pp1[f'impp1_MOZ_FREDDY_{time_case}'].rename(columns={'index_right': 'NAME_index'}, inplace=True)

        # Perform the spatial join as before
        jn_dfs = sjoin(disrup_pp1[f'impp1_MOZ_FREDDY_{time_case}'], disrupt_rd1[time_case], how='inner', predicate='intersects')

        # Assuming jn_dfs is your GeoDataFrame
        import geopandas as gpd

        # Group by 'geometry_orig' and sum the 'people' column for each group
        population_by_linestring = jn_dfs.groupby('geometry_orig')['imp_people'].sum().reset_index()

        # Convert the result back to a GeoDataFrame if you need to perform spatial operations
        population_by_linestring_gdf = gpd.GeoDataFrame(population_by_linestring, geometry='geometry_orig')

        # Save the result in the dictionary with the time_case as the key
        results_dict[time_case] = population_by_linestring_gdf

    return results_dict
```

```python
# IMPACT PLOTTING THEREFORE WE NEED TO EXTRACT IT LONG,LAT,LEVELS. 
## ONCE ALL THE CALCULATION ARE DONE IN DATAFRAMES AND IN EXPOSURES, SUBSET ONLY THE TRACKS TO PLOT ACCORDING TO THE LEAD TIME FORECAST
## AND THE DAY THAT MAKES LANDFALL, 
from datetime import datetime
import pandas as pd

def ext_col_categories(exp_instance, date_forecast):
    try:
        matching_columns = [col for col in exp_instance.gdf.columns if col.endswith(date_forecast)]
        if not matching_columns:
            print(f"No column ending with {date_forecast} found.")
            return None, None

        combined_data = pd.DataFrame()
        for impact_column in matching_columns:
            impact_data = exp_instance.gdf[['latitude', 'longitude', impact_column]].copy()
            impact_data.rename(columns={impact_column: 'impact'}, inplace=True)
            combined_data = pd.concat([combined_data, impact_data])

        combined_data.drop_duplicates(subset=['latitude', 'longitude'], inplace=True)
        levels = exp_instance.gdf['levels'] if 'levels' in exp_instance.gdf.columns else None

    except KeyError as ke:
        print(f"Error: {ke}. Could not extract data. Check column names and data structure.")
        return None, None
    
    return combined_data, levels
    
 from datetime import datetime
import pandas as pd
def extract_points_to_plot(exposures_dict, exp_key, date_forecast, event_date):
    try:
        dt_forecast = datetime.strptime(date_forecast, "%Y%m%d%H%M%S")
        dt_event = datetime.strptime(event_date, "%Y%m%d%H%M%S")
        lead_time = (dt_event - dt_forecast).days
        if lead_time < 0:
            print(f"Warning: The event date {event_date} is earlier than the forecast date {date_forecast}.")
        elif lead_time == 0:
            print(f"Note: The event date and forecast date are the same.")
        else:
            print(f"Note: Calculating with lead time of {lead_time} days.")

        exp_instance = exposures_dict[exp_key]
        combined_impact_data, exp_levels = ext_col_categories(exp_instance, date_forecast)
        
        # Initialize as None in case of any errors
        exp_lon, exp_lat, exp_impact, exp_level_data = None, None, None, None
        
        if combined_impact_data is not None:
            exp_impact = combined_impact_data[['impact', 'latitude', 'longitude']]
            exp_lon = combined_impact_data['longitude']
            exp_lat = combined_impact_data['latitude']
        
        if exp_levels is not None:
            exp_level_data = exp_instance.gdf[['levels', 'latitude', 'longitude']]

    except ValueError as ve:
        print(f"Error: {ve}. Ensure date_forecast and event_date are correctly formatted.")
        return None
    except KeyError as ke:
        print(f"Error: {ke}. Key not found in exposures_dict.")
        return None

    return {
        'longitude': exp_lon,
        'latitude': exp_lat,
        'impact_data': exp_impact,
        'level_data': exp_level_data,
        'lead_time': lead_time
    }
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Sample data - replace with actual data
data1 = tc_names['FREDDY_20230216180000'].intensity.data
data2 = tc_names['FREDDY_20230217180000'].intensity.data
data3 = tc_names['FREDDY_20230218180000'].intensity.data
data4 = tc_names['FREDDY_20230219180000'].intensity.data
data5 = tc_names['FREDDY_20230220180000'].intensity.data
data6 = tc_names['FREDDY_20230221180000'].intensity.data


# Calculating quantiles, mean, and max for each data set
stats = []
max_vals = []
for data in [data1, data2, data3, data4, data5, data6]:
    quantile_25 = np.quantile(data, 0.25)
    quantile_50 = np.quantile(data, 0.50)  # Median
    quantile_75 = np.quantile(data, 0.75)
    mean_val = np.mean(data)
    max_val = np.max(data)
    stats.append((quantile_25, quantile_50, quantile_75, mean_val))
    max_vals.append(max_val)

# Create a single bar plot with adjusted lines
fig, ax = plt.subplots(figsize=(10, 6))

# Dates for x-axis
dates = ['5 days', '4 days', '3 days', '2 days', '1 day', 'landfall']

# Adding bars for max values
bars = ax.bar(dates, max_vals, color='purple', alpha=0.7)

# Function to calculate the x position for lines
def line_xpos(bar):
    bar_width = bar.get_width()
    bar_x = bar.get_x()
    return bar_x, bar_x + bar_width

# Adding lines for quantiles and mean within each bar
for i, (bar, stat) in enumerate(zip(bars, stats)):
    xmin, xmax = line_xpos(bar)
    ax.hlines(y=stat[0], xmin=xmin, xmax=xmax, color='green', linestyle='--', label='25th Percentile' if i == 0 else "")
    ax.hlines(y=stat[1], xmin=xmin, xmax=xmax, color='blue', linestyle='--', label='Median (50th Percentile)' if i == 0 else "")
    ax.hlines(y=stat[2], xmin=xmin, xmax=xmax, color='red', linestyle='--', label='75th Percentile' if i == 0 else "")
    ax.hlines(y=stat[3], xmin=xmin, xmax=xmax, color='purple', linestyle='--', label='Mean' if i == 0 else "")

# Adding labels and title
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
ax.set_ylabel('Intensity (m/s)', fontsize=20)
ax.set_xlabel('Lead time', fontsize=20)
ax.set_title('')
ax.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize=15)

# Saving the figure as a JPG file
file_path = '/cluster/project/climate/gespejo/MDG_plots/Intensity_Plot_stat_16022023.jpg'  # Replace with your actual file path
plt.savefig(file_path, format='jpg', dpi=300)

plt.show()
```
