```python
import numpy as np
import pandas as pd

hazard_stats = {
    'lead_time': [],
    'mean': [],
    'std': [],
    'min': [],
    'max': [],
    'cv_hazard': []  # Coefficient of Variation
}

for key, tc_obj in new_tc_names.items():  # tc_obj is a TropCyclone object
    ensemble_values = []

    for i in range(51):  # Ensemble members 0 to 50
        try:
            data = tc_obj.intensity[i].data  # Get wind intensity data
            if isinstance(data, np.ndarray):
                flat_data = data.flatten()
                flat_data = flat_data[~np.isnan(flat_data)]  # remove NaNs
                if flat_data.size > 0:
                    ensemble_values.append(flat_data)
        except (IndexError, AttributeError, TypeError) as e:
            print(f"Skipping member {i} for {key}: {e}")
            continue

    if ensemble_values:
        all_data = np.concatenate(ensemble_values)
        mean_val = np.mean(all_data)
        std_val = np.std(all_data)
        min_val = np.min(all_data)
        max_val = np.max(all_data)
        cv_val = std_val / mean_val if mean_val > 0 else np.nan

        hazard_stats['lead_time'].append(key)
        hazard_stats['mean'].append(mean_val)
        hazard_stats['std'].append(std_val)
        hazard_stats['min'].append(min_val)
        hazard_stats['max'].append(max_val)
        hazard_stats['cv_hazard'].append(cv_val)
    else:
        print(f"No valid data for {key}, skipping.")

# Convert to DataFrame
df_hazard_stats = pd.DataFrame(hazard_stats)

# Optional: convert and sort by datetime if the key format includes a timestamp
try:
    df_hazard_stats['lead_time_dt'] = pd.to_datetime(
        df_hazard_stats['lead_time'].str.extract(r'(\d{8,})')[0], format='%Y%m%d%H%M%S')
    df_hazard_stats = df_hazard_stats.sort_values('lead_time_dt')
except Exception as e:
    print(f"Date parsing failed: {e}")

# Display result
print(df_hazard_stats)
```

```python
# Calculate CVs based on actual imph2_impp1 data

cv_table = {}

for lead_time, df_lead in imph2_impp1.items():
    mean_pp = df_lead['pp_affected'].mean()
    std_pp = df_lead['pp_affected'].std()
    cv_pp = std_pp / mean_pp if mean_pp > 0 else None

    mean_ens = df_lead['n_ensemble_members'].mean()
    std_ens = df_lead['n_ensemble_members'].std()
    cv_ens = std_ens / mean_ens if mean_ens > 0 else None

    cv_table[lead_time] = {
        'cv_pp_affected': cv_pp,
        'cv_n_ensemble_members': cv_ens
    }

# Convert to DataFrame
cv_df_actual = pd.DataFrame.from_dict(cv_table, orient='index')
cv_df_actual.index.name = 'lead_time'
cv_df_actual.reset_index(inplace=True)
```

```python
cv_df_actual
```

```python
# Merge cv_imph2 from df_impact_stats_imph2
cv_df_actual = cv_df_actual.merge(
    df_impact_stats_imph2[['lead_time', 'cv_imph2']],
    on='lead_time',
    how='left'
)

# Merge cv_imph1 from df_impact_stats_imph1
cv_df_actual = cv_df_actual.merge(
    df_impact_stats_imph1[['lead_time', 'cv_imph1']],
    on='lead_time',
    how='left'
)

# Merge cv_hazard from df_hazard_stats
cv_df_actual = cv_df_actual.merge(
    df_impact_stats_imph1_cat[['lead_time', 'cv_imph1_cat']],
    on='lead_time',
    how='left'
)

# Merge cv_hazard from df_hazard_stats
cv_df_actual = cv_df_actual.merge(
    df_hazard_stats[['lead_time', 'cv_hazard']],
    on='lead_time',
    how='left'
)
```

```python
cv_df_actual
```

```python
# Normalize each CV column to range [0, 1] independently
cv_df_normalized = cv_df_actual.copy()

# Columns to normalize
cv_columns = ['cv_pp_affected', 'cv_n_ensemble_members', 'cv_imph2', 'cv_imph1', 'cv_imph1_cat', 'cv_hazard']

# Normalize: (x - min) / (max - min)
for col in cv_columns:
    min_val = cv_df_normalized[col].min()
    max_val = cv_df_normalized[col].max()
    cv_df_normalized[f'{col}_norm'] = (cv_df_normalized[col] - min_val) / (max_val - min_val)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Prepare data for heatmap
heatmap_data = cv_df_normalized.set_index('lead_time')[[
    'cv_hazard_norm', 'cv_imph1_norm','cv_imph2_norm','cv_imph1_cat_norm', 'cv_n_ensemble_members_norm'
]].T

# Rename rows for clarity
heatmap_data.index = ['Hazard', 'Impact (Eberenz)','Impact (cat-1)' ,'Impact (Deltares)', 'Ensemble Agreement']

# Create heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(heatmap_data, annot=True, cmap='YlOrRd', cbar_kws={'label': 'Normalized Uncertainty (CV)'}, linewidths=0.5)

plt.title('Uncertainty Across Forecast Stages and Lead Times (Normalized CV)')
plt.xlabel('Lead Time')
plt.ylabel('Forecast Stage')
plt.tight_layout()
plt.show()
```
