```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.cm as cm_mp

# Definitions
SAFFIR_SIM_CAT = [34, 64, 83, 96, 113, 137, 1000]
CAT_NAMES = {
    -1: 'Tropical Depression',
    0: 'Tropical Storm',
    1: 'Hurricane Cat. 1',
    2: 'Hurricane Cat. 2',
    3: 'Hurricane Cat. 3',
    4: 'Hurricane Cat. 4',
    5: 'Hurricane Cat. 5',
}

# Generate colors
CAT_COLORS = cm_mp.rainbow(np.linspace(0, 1, len(SAFFIR_SIM_CAT)))

# Plotting dummy lines to generate legends
plt.figure(figsize=(10, 10))
for i, color in enumerate(CAT_COLORS, start=-1):
    plt.plot([], [], color=color, label=CAT_NAMES.get(i, 'Unknown'), lw=4)

# Create the legend
plt.legend(title='Saffir-Simpson Hurricane Wind Scale', loc='center', frameon=False)

# Hide axes
plt.gca().set_axis_off()

# Save the legend as a figure
legend_path = '/cluster/project/climate/gespejo/MDG_plots/saffir_simpson_legend_lines.jpg'
plt.savefig(legend_path, format='jpg', dpi=300)
plt.show()
# Return the path to the saved file
legend_path
```
