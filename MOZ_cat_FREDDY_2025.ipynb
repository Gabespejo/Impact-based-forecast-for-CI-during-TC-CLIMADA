```python
from datetime import datetime, timedelta
import os
import pandas as pd
import numpy as np
from climada_petals.hazard import TCForecast

def fetch_multiple_ecmwf_data_bufr4(base_folder, start_date, end_date, event_name):
    current_date = start_date
    forecast_variables = {}  # Dictionary to store forecast variables

    while current_date <= end_date:
        for utc_hour in [12]:  # Only fetch 12 UTC (ignoring 0 UTC)
            date_folder = current_date.replace(hour=utc_hour, minute=0, second=0).strftime("%Y%m%d%H%M%S")
            folder_path = os.path.join(base_folder, date_folder)

            if os.path.exists(folder_path):
                forecast = TCForecast()
                files_to_fetch = []
                for root, dirs, files in os.walk(folder_path):
                    for file in files:
                        if event_name in file:
                            files_to_fetch.append(os.path.join(root, file))

                if files_to_fetch:
                    print(f"Fetching data for {date_folder}")
                    forecast.fetch_ecmwf(files=files_to_fetch)
                    
                    # Convert `run_datetime` to a `pandas.Timestamp` with UTC timezone
                    for track in forecast.data:
                        if 'run_datetime' in track.attrs:
                            run_datetime = track.attrs['run_datetime']

                            # Convert numpy.datetime64 to pandas.Timestamp with UTC
                            if isinstance(run_datetime, np.datetime64):
                                run_datetime = pd.Timestamp(run_datetime).tz_localize("UTC")
                            
                            # Convert if it is a string (sometimes stored as string)
                            elif isinstance(run_datetime, str):
                                run_datetime = pd.Timestamp(run_datetime).tz_localize("UTC")

                            # Convert naive datetime to UTC timestamp
                            elif isinstance(run_datetime, datetime) and run_datetime.tzinfo is None:
                                run_datetime = pd.Timestamp(run_datetime).tz_localize("UTC")

                            # Ensure it's stored back in the track
                            track.attrs['run_datetime'] = run_datetime

                    variable_name = f"{event_name}_{date_folder}"
                    forecast_variables[variable_name] = forecast
                else:
                    print(f"No data found for {date_folder}")
            else:
                print(f"Folder not found: {date_folder}")

        current_date += timedelta(days=1)

    return forecast_variables
```

```python
def _find_duplicates(overlap_gdf):
    indices_dupl = np.unique([index_right for index_right, index_left in zip(overlap_gdf.index_right, overlap_gdf.index_left)
                    if index_right != index_left])
    indices_dupl = indices_dupl[:int(len(indices_dupl/2))] # only first half of duplicates should be removed!
    return indices_dupl

def _remove_duplicates(orig_gdf, indices_dupl):
    indices_keep = list(set(orig_gdf.index).difference(set(indices_dupl)))
    return orig_gdf.loc[indices_keep]

def find_remove_duplicates(orig_gdf):
    if 'index' in orig_gdf.columns:
        orig_gdf = orig_gdf.drop('index', axis=1)
    overlap_gdf = orig_gdf.sjoin(orig_gdf)
    if 'index_left' not in overlap_gdf.columns:
        overlap_gdf = overlap_gdf.reset_index()
        overlap_gdf = overlap_gdf.rename({'index':'index_left'}, axis=1)  
    indices_dupl = _find_duplicates(overlap_gdf)
    return _remove_duplicates(orig_gdf, indices_dupl) 
```

```python
def plot_cat_tracks_without_levels(tc_tracks, exp_lon, exp_lat, exp_value1, exp_value2, exp_value3, figure_filename,event_name):
    
    #######################################################
    color = ['#01B4CC', '#0000ff','#FFD700', '#B8860B', '#CD6600', '#8B4500', '#CD3333']
    CAT_COLORS = color[0:]
    extent = (exp_lon.min(), exp_lon.max(), exp_lat.min(), exp_lat.max())
    mid_lon = 0.5 * (extent[1] + extent[0])
    proj = ccrs.PlateCarree()
    _, axis, fontsize = u_plot.make_map(proj=proj, figsize=(8, 8), adapt_fontsize=True)
    axis.set_extent(extent, crs=ccrs.PlateCarree())
    #u_plot.add_shapes(axis)
    cmap = ListedColormap(colors=CAT_COLORS)
    norm1 = BoundaryNorm([0] + SAFFIR_SIM_CAT, len(SAFFIR_SIM_CAT))
    for track in tc_tracks.data:
        lonlat = np.stack([track.lon.values, track.lat.values], axis=-1)
        lonlat[:, 0] = u_coord.lon_normalize(lonlat[:, 0], center=mid_lon)
        segments = np.stack([lonlat[:-1], lonlat[1:]], axis=1)
        segments = segments[segments[:, 0, 0] * segments[:, 1, 0] >= 0, :, :]
        track_lc = LineCollection(segments, cmap=cmap, norm=norm1, linestyle=':', alpha=0.3)
        track_lc.set_array(track.max_sustained_wind.values)
        axis.add_collection(track_lc)
        mask = (segments[:, 0, 0] > 100) & (segments[:, 1, 0] < -100)
        segments[mask, 1, 0] = 180
        mask = (segments[:, 0, 0] < -100) & (segments[:, 1, 0] > 100)
        segments[mask, 1, 0] = -180
    ## Lead time 
    lead_time_str = f"{lead_time} days"
    
    ##### 
    gadm_MOZ.plot(ax=axis, facecolor='none', edgecolor='gray', linewidth=0.5)
    
    # Create a color normalization object
    
    color_for_exp_value1 = (204 / 255, 255 / 255, 102 / 255, 0.9)  # Corresponds to a light green
    color_for_exp_value2 = (255 / 255, 153 / 255, 0 / 255, 0.9)    # Corresponds to orange
    color_for_exp_value3 = (178 / 255, 34 / 255, 34 / 255, 0.9) 

    # Combine all data into a single DataFrame for processing
    all_data = pd.concat([
        exp_impact1.assign(source='exp_value1'),
        exp_impact2.assign(source='exp_value2'),
        exp_impact3.assign(source='exp_value3')
    ])

    # Sort the data by 'source' to enforce the hierarchy: exp_value3 > exp_value2 > exp_value1
    all_data_sorted = all_data.sort_values('source', ascending=False)

    # Group by longitude and latitude
    grouped = all_data_sorted.groupby(['longitude', 'latitude'])

    # Initialize an empty DataFrame to store the results
    selected_data = []

    # Iterate through each group
    for (lon, lat), group in grouped:
        selected_row = None
        # Default to exp_value1
        #selected_row = group[group['source'] == 'exp_value1'].iloc[0]
        
        # Check if exp_value1's impact is >= 0.5
        if any(group[group['source'] == 'exp_value1']['impact'] >= 0.1):
            selected_row = group[group['source'] == 'exp_value1'].iloc[0]
        
        # Check if exp_value2's impact is >= 0.5
        if any(group[group['source'] == 'exp_value2']['impact'] >= 0.1):
            selected_row = group[group['source'] == 'exp_value2'].iloc[0]
    
        # Check if exp_value3's impact is >= 0.5
        if any(group[group['source'] == 'exp_value3']['impact'] >= 0.1):
            selected_row = group[group['source'] == 'exp_value3'].iloc[0]
    
        # Append the selected row to the list
        #selected_data.append(selected_row)
        # Append the selected row to the list if any condition is met
        if selected_row is not None:
            selected_data.append(selected_row)

        # Convert the list to a DataFrame
        unique_locations = pd.DataFrame(selected_data)

        # Reset the index of the DataFrame
        unique_locations.reset_index(drop=True, inplace=True)

    # Create a ListedColormap with the defined colors
    cmap = ListedColormap([color_for_exp_value3, color_for_exp_value2, color_for_exp_value1])
    norm = plt.Normalize(1, 3)

    # Now plot using mappable
    for _, row in unique_locations.iterrows():
        if row['impact'] > 0:
            source_index = {'exp_value1': 3, 'exp_value2': 2, 'exp_value3': 1}[row['source']]
            mappable = axis.scatter(row['longitude'], row['latitude'], s=30, c=[source_index], cmap=cmap, norm=norm, label=row['source'], 
                                    marker='o',zorder=3)
    
    # Create a legend for the color ranges and position it on the right
    color_legend = axis.legend(handles=legend_patches, loc='upper right', bbox_to_anchor=(-0.2, 0.2), fontsize=15, 
                               title='Hospitals in Risk by Categories')
    color_legend.get_title().set_fontsize('15')  # Set the font size for the legend title
    # Add event_name at the top of the legend

    # Create a new legend for the ensemble forecast tracks
    # Create a Line2D element for the ensemble forecast tracks (dashed lines)
    leg_lines = [Line2D([0], [0], color=CAT_COLORS[i_col], lw=2,linestyle=':')
             for i_col in range(len(SAFFIR_SIM_CAT))]
    leg_names = [CAT_NAMES[i_col] for i_col in range(len(SAFFIR_SIM_CAT)) if i_col in CAT_NAMES]

    axis.legend(leg_lines, leg_names, loc='upper left', bbox_to_anchor=(-0.9, 0.6), fontsize=15,
                title='51 ensembe members',title_fontsize=15)
    

    # Add all three legends to the axis
    axis.add_artist(color_legend)

    # Set the figure size to accommodate the legends
    plt.gcf().set_size_inches(14,8)
   
    ### run_date time 
    lead_time_str = f"{lead_time} days"
    run_datetime = tc_tracks.data[0].attrs['run_datetime']
    # Assuming run_datetime is in the desired format ('%a %d %b %Y %HUTC') and lead_time is in days
    event_time = run_datetime + timedelta(days=lead_time)

    # Format event_time similarly to run_datetime
    event_time_str = event_time.strftime('%a %d %b %Y %HUTC')
    
    # Assuming run_datetime is in the desired format ('%a %d %b %Y %HUTC') and lead_time is in days
    forecast_date_str = "Forecast date: " + run_datetime.strftime('%a %d %b %Y %HUTC')
    # Add the forecast date text to the figure
    plt.figtext(0.01, 1.1, forecast_date_str, color='red', fontsize=22, ha='left')
    ###changuin the values of 0.145,0.09, 0.9 changes the location of the datetime 
    
    # Add the event time text to the figure
    event_text = f'{event_name} - Landfall on: {event_time_str}'
    plt.figtext(0.4, 1.05, event_text, color='red', fontsize=18, ha='left')
    plt.figtext(0.4, 1,'Lead time: ' +lead_time_str,color='k', fontsize=18, ha='left')
    plt.tight_layout()
    # Set the x-axis (longitude) and y-axis (latitude) limits
    #plt.xlim(lon_min, lon_max)
    #plt.ylim(-24, -15)
    # Save the figure to a file
    plt.savefig(figure_filename, dpi=300,bbox_inches='tight',facecolor='white')
    plt.show()
```

```python
def  plot_cat_tracks_withlevels(tc_tracks, exp_lon, exp_lat, exp_impact1, exp_impact2, exp_impact3, exp_level, figure_filename, event_name):
    #######################################################
    color = ['#01B4CC', '#0000ff','#FFD700', '#B8860B', '#CD6600', '#8B4500', '#CD3333']
    CAT_COLORS = color[0:]
    extent = (exp_lon.min(), exp_lon.max(), exp_lat.min(), exp_lat.max())
    mid_lon = 0.5 * (extent[1] + extent[0])
    proj = ccrs.PlateCarree()
    _, axis, fontsize = u_plot.make_map(proj=proj, figsize=(8, 8), adapt_fontsize=True)
    axis.set_extent(extent, crs=ccrs.PlateCarree())

    cmap = ListedColormap(colors=CAT_COLORS)
    norm1 = BoundaryNorm([0] + SAFFIR_SIM_CAT, len(SAFFIR_SIM_CAT))
    for track in tc_tracks.data:
        lonlat = np.stack([track.lon.values, track.lat.values], axis=-1)
        lonlat[:, 0] = u_coord.lon_normalize(lonlat[:, 0], center=mid_lon)
        segments = np.stack([lonlat[:-1], lonlat[1:]], axis=1)
        segments = segments[segments[:, 0, 0] * segments[:, 1, 0] >= 0, :, :]
        track_lc = LineCollection(segments, cmap=cmap, norm=norm1, linestyle=':', alpha=0.3)
        track_lc.set_array(track.max_sustained_wind.values)
        axis.add_collection(track_lc)

    gadm_MOZ.plot(ax=axis, facecolor='none', edgecolor='gray', linewidth=0.5)

    # Define colors for exp_value levels
    color_for_exp_value1 = (204 / 255, 255 / 255, 102 / 255, 0.9)  # Light green
    color_for_exp_value2 = (255 / 255, 153 / 255, 0 / 255, 0.9)    # Orange
    color_for_exp_value3 = (178 / 255, 34 / 255, 34 / 255, 0.9)    # Red

    # Combine all data into a single DataFrame for processing
    all_data = pd.concat([
        exp_impact1.assign(source='exp_value1'),
        exp_impact2.assign(source='exp_value2'),
        exp_impact3.assign(source='exp_value3')
    ])

    # Sort by source: exp_value3 > exp_value2 > exp_value1
    all_data_sorted = all_data.sort_values('source', ascending=False)
    grouped = all_data_sorted.groupby(['longitude', 'latitude'])
    selected_data = []

    # Filter based on hierarchy and impact threshold
    for (lon, lat), group in grouped:
        selected_row = None
        if any(group[group['source'] == 'exp_value1']['impact'] >= 0.1):
            selected_row = group[group['source'] == 'exp_value1'].iloc[0]
        if any(group[group['source'] == 'exp_value2']['impact'] >= 0.1):
            selected_row = group[group['source'] == 'exp_value2'].iloc[0]
        if any(group[group['source'] == 'exp_value3']['impact'] >= 0.1):
            selected_row = group[group['source'] == 'exp_value3'].iloc[0]

        if selected_row is not None:
            selected_data.append(selected_row)

    unique_locations = pd.DataFrame(selected_data)
    unique_locations.reset_index(drop=True, inplace=True)

    # Create a ListedColormap
    cmap = ListedColormap([color_for_exp_value3, color_for_exp_value2, color_for_exp_value1])
    norm = plt.Normalize(1, 3)

    # Plot the selected impact points
    for _, row in unique_locations.iterrows():
        if row['impact'] > 0:
            source_index = {'exp_value1': 3, 'exp_value2': 2, 'exp_value3': 1}[row['source']]
            axis.scatter(row['longitude'], row['latitude'], s=10, c=[source_index], cmap=cmap, norm=norm, label=row['source'],
                         marker='o', zorder=4)

    #######################################################
    ###  NEW CODE BLOCK: 
    #######################################################
    for level, marker in level_markers.items():
        # Filter data for the current level
        level_data = exp_level[exp_level['levels'] == level]

        # Get style settings for this level
        style = level_style.get(level, {'alpha': 0.2, 'edgecolor': 'black', 'facecolor': 'none', 'size': 10})

        # Plot markers with specified style and size
        axis.scatter(
            level_data['longitude'],
            level_data['latitude'],
            marker=marker,
            s=style['size'],
            label=f'Level {level}',
            edgecolor=style['edgecolor'],
            facecolor=style['facecolor'],
            linewidths=0.3,
            alpha=style['alpha'],
            zorder=2  # Ensure these markers appear on top
        )

    #######################################################
    ### Add Forecast Date, Event Name & Save Plot
    #######################################################
    lead_time_str = f"{lead_time} days"
    run_datetime = tc_tracks.data[0].attrs['run_datetime']
    event_time = run_datetime + timedelta(days=lead_time)
    event_time_str = event_time.strftime('%a %d %b %Y %HUTC')
    forecast_date_str = "Forecast date: " + run_datetime.strftime('%a %d %b %Y %HUTC')

    plt.figtext(0.2, 1.1, forecast_date_str, color='red', fontsize=18, ha='left')
    event_text = f'{event_name} - Landfall on: {event_time_str}'
    plt.figtext(0.2, 1.05, event_text, color='red', fontsize=18, ha='left')
    plt.figtext(0.4, 1, 'Lead time: ' + lead_time_str, color='k', fontsize=18, ha='left')

    plt.gcf().set_size_inches(14, 8)
    plt.tight_layout()
    plt.savefig(figure_filename, dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()
```

```python
import geopandas as gpd
from geopandas.tools import sjoin

def process_time_cases(disrupt_rd1, disrup_pp1):
    # Buffer distance for 2 km in degrees
    buffer_distance_2km_deg = 2 / 111  # 2 km converted to degrees

    # Dictionary to store results
    results_dict = {}

    # Assuming your data is already loaded into disrupt_rd1 and disrup_pp1
    for time_case in disrupt_rd1.keys():
        # 1. Buffer the Linestrings
        disrupt_rd1[time_case]['buffered_geometry'] = disrupt_rd1[time_case].geometry.buffer(buffer_distance_2km_deg)

        # Set the buffered geometry as the active geometry
        disrupt_rd1[time_case].set_geometry('buffered_geometry', inplace=True)

        # Rename the existing 'index_right' column to avoid conflict
        disrup_pp1[f'{time_case}'].rename(columns={'index_right': 'NAME_index'}, inplace=True)

        # Perform the spatial join as before
        jn_dfs = sjoin(disrup_pp1[f'{time_case}'], disrupt_rd1[time_case], how='inner', predicate='intersects')

        # Assuming jn_dfs is your GeoDataFrame
        import geopandas as gpd

        # Group by 'geometry_orig' and sum the 'people' column for each group
        population_by_linestring = jn_dfs.groupby('geometry_orig').agg({
        'imp_people': 'sum',  # summing up the people
        'longitude': 'first',  # keeping the first longitude value
        'latitude': 'first',  # keeping the first latitude value
        'geometry': 'first',  # keeping the first geometry value
        'index_right': 'first',  # keeping the first index_right value
        'osm_id': 'first',  # keeping the first osm_id value
        'mean_impact': 'first'  # keeping the first mean_impact value
        }).reset_index()
        # Convert the result back to a GeoDataFrame if you need to perform spatial operations
        population_by_linestring_gdf = gpd.GeoDataFrame(population_by_linestring, geometry='geometry_orig')

        # Save the result in the dictionary with the time_case as the key
        results_dict[time_case] = population_by_linestring_gdf

    return results_dict
```
